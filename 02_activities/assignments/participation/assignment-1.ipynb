```python
# load libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy 
import PIL
import requests
```

```python
#setting a seed for randomization / reproducibility
np.random.seed(613)
#x-axis, generate an array of values from 0:50
x = np.arange(50)
#y-axis, generate random numbers from a range
y = np.random.randint(0, 100, 50)

```

```python
#create a chart as a scatterplot 
fig, ax = plt.subplots(figsize = (5,3))
ax.scatter(x, y)
```

```python
#same chart but as a bargraph 
fig, ax = plt.subplots(figsize=(5, 3))
ax.bar(x,y)
```

```python
#same chart as a line plot
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
```

```python
#same chart as a histogram (x is omitted)
fig, ax = plt.subplots(figsize=(5, 3))
ax.hist(y)
```

```python
#same line graph as before
fig, ax = plt.subplots(figsize = (5,3))
ax.plot(x,y)

```

```python
ax.set_title('Total growth over time')
ax.set_ylabel('Total growth')
ax.set_xlabel('Years since start')
```

```python
#same line graph as before but with labels
fig, ax = plt.subplots(figsize = (5,3))
ax.plot(x,y)
ax.set_title('Total growth over time')
ax.set_ylabel('Total growth')
ax.set_xlabel('Years since start')
```

```python
#creating dictionaries for modified fonts (fonts 1/2 for different elements)
font1 = {'family':'sans-serif','color':'red','size':20}

font2 = {'family':'monospace','color':'blue',
         'size':14}
```

```python
#add fontdict with modified fonts, different locations of title/labels . tight layout (?)
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y)

ax.set_title('Total growth over time', fontdict = font1, loc='center')
ax.set_ylabel('Total growth', fontdict = font2, loc='bottom')
ax.set_xlabel('Years since start', fontdict = font2, loc='left')
fig.tight_layout()
```

```python
#make a choropleth map
# Import the geopandas and geoplot libraries
import geopandas as gpd
import geoplot.crs as gcrs
import geoplot as gplt
import matplotlib.pyplot as plt
import seaborn as sns
import cartopy.crs as ccrs

```

```python
# Load the json file with county coordinates
geoData = gpd.read_file(
    'https://raw.githubusercontent.com/holtzy/The-Python-Graph-Gallery/master/static/data/US-counties.geojson'
)

# Make sure the "id" column is an integer
geoData.id = geoData['id'].astype(int)

# Remove Alaska, Hawaii and Puerto Rico.
statesToRemove = ['02', '15', '72']
geoData = geoData[~geoData.STATE.isin(statesToRemove)]

# Basic plot with just county outlines
gplt.polyplot(
    geoData,
    projection=gcrs.AlbersEqualArea()
)

# Read file
data = pd.read_csv(
    'https://raw.githubusercontent.com/holtzy/The-Python-Graph-Gallery/master/static/data/unemployment-x.csv')

# Show the distribution of unemployment rate
sns.histplot(data["rate"])

fullData = geoData.merge(
    data,
    left_on=['id'],  # identifier from geodataframe
    right_on=['id']  # identifier from dataframe
)
fullData.head(3)

# for color mapping with 10 different colors
import mapclassify as mc
scheme = mc.Quantiles(fullData['rate'], k=10)

# Map
gplt.choropleth(
    fullData,
    projection=gcrs.AlbersEqualArea(),
    hue="rate",
    scheme=scheme, cmap='inferno_r',
    linewidth=.1,
    edgecolor='black',
    figsize=(12, 8)
)
plt.show()
```

```python
# load libraries
import numpy as np 
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests

# sample data
np.random.seed(613)
x = np.arange(50)
y1 = np.random.randint(0,100,50)
y2 = np.random.randint(0,100,50)
```

```python

# line plot with both y variables on same axes

fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y1)
ax.plot(x,y2)
```

```python

# adding a legend (upper, lower, center; left, center, right)

fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y1, label = 'Person 1')
ax.plot(x,y2, label = 'Person 2')
ax.legend(loc = 'upper right')
```

```python
#Adding frame, font size and columns, legend shadow 
 
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y1, label = 'Person 1')
ax.plot(x,y2, label = 'Person 2')
ax.legend(loc = 'lower center',
          frameon = True,
          fontsize = 10,
          ncol = 2, 
          shadow = True)
```

```python
#Adding bbox_to_anchor
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y1, label = 'Person 1')
ax.plot(x,y2, label = 'Person 2')
ax.legend(loc = 'upper right',
          bbox_to_anchor = (0.1,0.1),
          fontsize = 10,
          ncol = 2, 
          shadow = True)
```

```python
#Adding bbox_to_anchor
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y1, label = 'Person 1')
ax.plot(x,y2, label = 'Person 2')
ax.legend(loc = 'lower right',
          bbox_to_anchor = (1,1))
```

```python
# Adding annotations

fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y1, label ="Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')

ax.text(12,80,"This value is important!",
        ha = 'center',
        color = 'red',
        size = 12)
```

```python
# Transform: Empty axes 
fig, ax = plt.subplots()
ax.axis([0,10,0,10])

```

```python
# Transform: annotations to data
fig, ax = plt.subplots()
ax.axis([0,10,0,10])

ax.text(1,7, ". Data:(1,7)", transform=ax.transData)
```

```python
# Transform: to axes 
fig, ax = plt.subplots()
ax.axis([0,10,0,10])
ax.text(0.6, 0.6, ". Axes:(0.6, 0.6)", transform=ax.transAxes)
```

```python
#Transform: to figure
fig, ax = plt.subplots()
ax.axis([0,10,0,10])
ax.text(0.1, 0.1, ". Figure:(0.1., 0.1)", transform=fig.transFigure)
```

```python
# Adding arrows
fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y1, label = 'Person 1')
ax.scatter(x,y2, label = 'Person 2')
ax.legend(loc='lower right', 
          bbox_to_anchor = (1,1),)

ax.annotate('This is important!', xy=(15,60), xytext=(25,94),
            arrowprops=dict(arrowstyle = 'wedge', color = 'hotpink'))
```

```python
#Adding arrow styles
fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y1, label = 'Person 1')
ax.scatter(x,y2, label = 'Person 2')
ax.legend(loc='lower right', 
          bbox_to_anchor = (1,1),)

ax.annotate('This is important!', xy=(15,60), xytext=(25,94),
            arrowprops=dict(arrowstyle = 'fancy', color = 'red'))
```

```python
#Using locator to remove ticks and labels
fig, ax = plt.subplots(figsize = (5,3))
ax.scatter(x,y1,label ="person 1")
ax.scatter(x,y2,label = "person2")
ax.legend(loc='lower right',
          bbox_to_anchor = (1,1))

ax.yaxis.set_major_locator(plt.NullLocator())
ax.xaxis.set_major_locator(plt.NullLocator())
```

```python
#Using formatter to remove just the labels
fig, ax = plt.subplots(figsize = (5,3))
ax.scatter(x,y1,label ="person 1")
ax.scatter(x,y2,label = "person2")
ax.legend(loc='lower right',
          bbox_to_anchor = (1,1))

ax.xaxis.set_major_formatter(plt.NullFormatter())
ax.yaxis.set_major_formatter(plt.NullFormatter())
```

```python
#Limiting the number of ticks
fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y1, label="Person 1")
ax.scatter(x,y2, label="Person 2")
ax.legend(loc='lower right',
          bbox_to_anchor = (1,1))

ax.xaxis.set_major_locator(plt.MaxNLocator(2))
ax.yaxis.set_major_locator(plt.MaxNLocator(3))
```

```python
#Changing tick interval 
fig, ax = plt.subplots(figsize = (5,3))
ax.scatter(x,y1, label = "person 1")
ax.scatter(x, y2, label = "person 2")
ax.legend(loc='lower right',
          bbox_to_anchor = (1,1))

ax.xaxis.set_major_locator(plt.MultipleLocator(3))
ax.yaxis.set_major_locator(plt.MultipleLocator(10))
```

```python
#Ticks at an angle
fig, ax = plt.subplots(figsize = (5,3))
ax.scatter(x,y1, label = "person 1")
ax.scatter(x,y2,label = "person 2")
ax.legend(loc='lower right',
          bbox_to_anchor = (1,1))

plt.xticks(rotation = 45, ha='right')
plt.yticks(rotation = 45, ha='right')
```

```python
# Modifying axis fonts

fig, ax = plt.subplots(figsize = (5,3))
ax.scatter(x,y1, label = 'Person 1')
ax.scatter(x,y2, label = 'Person 2')
ax.legend(loc='lower right',
          bbox_to_anchor = (1,1))

font1 = {'family':'sans-serif', 'color':'green'}
plt.xlabel('Shiny New X Axis!', fontsize = 14, fontdict = font1)
```
