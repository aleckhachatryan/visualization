Appendix: Code 

```python
#Loading libraries

import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
#Creating a dataframe from the source file

df = pd.read_csv(r'F:\Documents\DSI\visualization\02_activities\assignments\PollinateTO.csv')
```

```python
#Creating a visualization

#Font dictionary
font1 = {'size':14}
font2 = {'size':10}

#Scatterplot
ax = df.plot(kind = 'scatter', x = 'YEAR_FUNDED', y = 'ESTIMATED_GARDEN_SIZE')
ax.set_title('PollinateTO Garden Size (2020-2024)', fontdict = font1, loc='center')
ax.set_xlabel('Year', fontdict = font2, loc='center')
ax.set_ylabel('Garden Size', fontdict = font2, loc='center')
ax.xaxis.set_major_locator(plt.MaxNLocator(5))

#Saving the visualization
plt.savefig('viz1.png')
```
