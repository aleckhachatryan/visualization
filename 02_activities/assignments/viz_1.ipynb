Visualization 1. 

This is a visualization of the general trends in the implementation of Toronto's PollinateTO program in 2020-2024, including the number and size of community pollinator gardens in each year and their share in the total values over five years. 

# What software did you use to create your data visualization?
Microsoft Excel. I use Excel for simple tasks such as data cleaning and descriptive statistics and visualizations. In this case, I performed a few simple operations such as summarizing PollinateTO garden numbers and sizes for each year and calculating the percentage shares of these values in the total.   

# Who is your intended audience? 
The intended audience is professionals and the general public interested in expanding the PollinateTO program. They will benefit from understanding the program trends and see gaps and opportunities.  
    
# What information or message are you trying to convey with your visualization? 
Based on City of Toronto (2025) data, the visualization suggests that while the number of PollinateTO gardens being registered has been more or less constant (~ 100 per year), their total size has been decreasing throughout the program. In 2020, the first year of observations, the total size of these gardens amounted to 8,693 units, while in 2024 it amounted to 1731 units (note: the dataset description does not specify whether these units are sq.ft, acres, or ha). It is important to understand why this is happening and address gaps to encourage biodiversity in Toronto. 
    
# What aspects of design did you consider when making your visualization? How did you apply them? With what elements of your plots? 
I was interested in placing two variables: number and size of gardens side by side to understand the main trends in program development. Since these variables are of a different magnitude (hundreds vs. thousands), I had to choose between using a logarithmic scale to equalize them or using percentages instead of absolute numbers. I chose the latter because it provides a cleaner visualization, although I understand that using percentages requires more explanations and may create more confusion. Another solution would be to create two separate graphs to show relevant trends and to explain the differences via text.    
    
# How did you ensure that your data visualizations are reproducible? If the tool you used to make your data visualization is not reproducible, how will this impact your data visualization? 
As we discussed in class, reproducibility means the ability to reproduce the method and process using the same data or code. In this case, the visualization is reproducible because the dataset is available online and the method can be described thus: 
1. Using the Excel sumif function, I summarized the number of gardens per year. 
2. Using the same function, I summarized the size of gardens per year. 
3. Finally, I calculated the percentage share of the five-year total for both variables. 
These calculations are provided to facilitate reproducibility. 
       
# How did you ensure that your data visualization is accessible?  
Based on the Government of Canada principles discussed in class, I tried to make the visualization concise and simple, with large sans-serif fonts, contrasting colours, and accompanied with a legend.  
    
# Who are the individuals and communities who might be impacted by your visualization?  
I think the only individuals who might be impacted are pollinator garden initiators and implementors. 
    
# How did you choose which features of your chosen dataset to include or exclude from your visualization? 
I wanted to make the visualization simple and straightforward. I also wanted to choose variables that could help identify a potential gap or problem, although not necessarily a solution.

The labour of those who implemented the pollinator gardens and supplied relevant data to the City, and those who administer and implement the PollinateTO program.
